```python
import xml.etree.ElementTree as ET

def crear_xml_en_blanco(ruta_archivo):
    # Crear el elemento raíz del archivo KML
    kml = ET.Element('kml', attrib={'xmlns': 'http://www.opengis.net/kml/2.2'})

    # Crear el elemento Documento
    document = ET.SubElement(kml, 'Document')

    # Agregar un nombre al documento
    name = ET.SubElement(document, 'name')
    name.text = 'Proyectos S/E'

    # Agregar una carpeta de proyectos
    folder = ET.SubElement(document, 'Folder')
    folder_name = ET.SubElement(folder, 'name')
    folder_name.text = 'Proyectos S/E'

    # Convertir el elemento KML a una cadena de texto
    kml_string = ET.tostring(kml, encoding='utf-8', method='xml')

    # Escribir la cadena de texto a un archivo
    with open(ruta_archivo, 'wb') as archivo:
        archivo.write(kml_string)

# Ruta del archivo XML en blanco
ruta_archivo = 'archivo_en_blanco.kml'

# Crear el archivo XML en blanco
crear_xml_en_blanco(ruta_archivo)
```

```python
def crear_esquema(nombre):
    # El parametro "nombre" tendra un codigo de numeros, donde el primer numero sera el numero de filas de la tabla de patios, el segundo el nro de filas de la tabla trafos y el tercero el numero de parrafos de la seccion otros.
    # por ejemplo, si el nombre es Amp_110, la tabla de patios tendra 1 fila, la tabla de trafos 1 fila y la seccion otros 0 parrafos. Por lo tanto, su esquema debe recibir los parametros de 1 patio, 1 trafo y 0 otros.
    largo_nombre = len(nombre)
    n_otros = int(nombre[largo_nombre-1])
    n_trafos = int(nombre[largo_nombre-2])
    n_patios = int(nombre[largo_nombre-3])


    print(f"n_patios: {n_patios}")
    print(f"n_trafos: {n_trafos}")
    print(f"n_otros: {n_otros}")

    esquema_patios = ""
    esquema_trafos = ""
    esquema_otros = ""


    esquema_base= f"""
        <ns0:Schema name="{nombre}" id="{nombre}">
            <ns0:SimpleField type="string" name="OBRA">
                <ns0:displayName>OBRA</ns0:displayName>
            </ns0:SimpleField>
            <ns0:SimpleField type="string" name="DECRETO">
                <ns0:displayName>DECRETO</ns0:displayName>
            </ns0:SimpleField>
            <ns0:SimpleField type="string" name="TIPO">
                <ns0:displayName>TIPO</ns0:displayName>
            </ns0:SimpleField>
            <ns0:SimpleField type="double" name="VI">
                <ns0:displayName>VI</ns0:displayName>
            </ns0:SimpleField>
            <ns0:SimpleField type="string" name="ENTRADA_OP">
                <ns0:displayName>ENTRADA_OP</ns0:displayName>
            </ns0:SimpleField>
            <ns0:SimpleField type="string" name="RESUMEN">
                <ns0:displayName>RESUMEN</ns0:displayName>
            </ns0:SimpleField>"""


    if n_patios != 0:
        for i in range(n_patios):
            #el esquema base de los patios posee los atributos tipo, tension, configuracion, conexiones y posiciones_disponibles
            esquema_patios += f"""
            <ns0:SimpleField type="string" name="patio{i+1}_tipo">
                <ns0:displayName>patio{i+1}_tipo</ns0:displayName>
            </ns0:SimpleField>
            <ns0:SimpleField type="string" name="patio{i+1}_tension">
                <ns0:displayName>patio{i+1}_tension</ns0:displayName>
            </ns0:SimpleField>
            <ns0:SimpleField type="string" name="patio{i+1}_configuracion">
                <ns0:displayName>patio{i+1}_configuracion</ns0:displayName>
            </ns0:SimpleField>
            <ns0:SimpleField type="string" name="patio{i+1}_conexiones">
                <ns0:displayName>patio{i+1}_conexiones</ns0:displayName>
            </ns0:SimpleField>
            <ns0:SimpleField type="string" name="patio{i+1}_posiciones_disponibles">
                <ns0:displayName>patio{i+1}_posiciones_disponibles</ns0:displayName>
            </ns0:SimpleField>"""
            
    if n_trafos != 0:
        for i in range((n_trafos)):
            #el esquema base de los trafos posee los atributos tipo, tension_trafo_reemplazado, tension_trafo_nvo y Capacidad
            esquema_trafos += f"""
            <ns0:SimpleField type="string" name="trafo{i+1}_tipo">
                <ns0:displayName>trafo{i+1}_tipo</ns0:displayName>
            </ns0:SimpleField>
            <ns0:SimpleField type="string" name="trafo{i+1}_tension_trafo_reemplazado">
                <ns0:displayName>trafo{i+1}_tension_trafo_reemplazado</ns0:displayName>
            </ns0:SimpleField>
            <ns0:SimpleField type="string" name="trafo{i+1}_tension_trafo_nvo">
                <ns0:displayName>trafo{i+1}_tension_trafo_nvo</ns0:displayName>
            </ns0:SimpleField>
            <ns0:SimpleField type="string" name="trafo{i+1}_Capacidad">
                <ns0:displayName>trafo{i+1}_Capacidad</ns0:displayName>
            </ns0:SimpleField>"""

    if n_otros != 0:
        for i in range(n_otros):
            #en elñ caso de otros, solo se imprimira el parrafo, por lo que solo entregamos el atributo parrafo
            esquema_otros += f"""
            <ns0:SimpleField type="string" name="parrafo{i+1}">
                <ns0:displayName>parrafo{i+1}</ns0:displayName>
            </ns0:SimpleField>"""


    esquema_base += esquema_patios
    esquema_base += esquema_trafos
    esquema_base += esquema_otros

    esquema_base += """</ns0:Schema>"""

    return esquema_base


print(crear_esquema("Amp_200"))


```

```python
def generar_tabla_patios(nombre):

    largo_nombre = len(nombre)
    n_patios = int(nombre[largo_nombre-3])


    estilo_base_patio = f"""
        &lt;table border="1" cellpadding="4" cellspacing="0" width="100%" style="margin-top: 10px; text-align: center;"&gt;
        &lt;tr&gt;
        &lt;th bgcolor="#AAAAAA"&gt;Tipo&lt;/th&gt;
        &lt;th bgcolor="#AAAAAA"&gt;Tensión&lt;/th&gt;
        &lt;th bgcolor="#AAAAAA"&gt;Configuración&lt;/th&gt;
        &lt;th bgcolor="#AAAAAA"&gt;Conexiones&lt;/th&gt;
        &lt;th bgcolor="#AAAAAA"&gt;Posiciones Disponibles&lt;/th&gt;
        &lt;/tr&gt;"""

    for i in range(n_patios):
        estilo_base_patio += f"""
        &lt;tr&gt;
        &lt;td&gt;$[{nombre}/patio{i+1}_tipo]&lt;/td&gt;
        &lt;td&gt;$[{nombre}/patio{i+1}_tension]&lt;/td&gt;
        &lt;td&gt;$[{nombre}/patio{i+1}_configuracion]&lt;/td&gt;
        &lt;td&gt;
            &lt;div style="overflow: auto; text-align: justify"&gt;
                $[{nombre}/patio{i+1}_conexiones]
            &lt;/div&gt;
        &lt;/td&gt;
        &lt;td&gt;$[{nombre}/patio{i+1}_posiciones_disponibles]&lt;/td&gt;
        &lt;/tr&gt;"""

    estilo_base_patio += """&lt;/table&gt;"""

    return estilo_base_patio

def generar_tabla_trafos(nombre):
    
    largo_nombre = len(nombre)
    n_trafos = int(nombre[largo_nombre-2])


    estilo_base_trafo = f"""
        &lt;table border="1" cellpadding="4" cellspacing="0" width="100%" style="margin-top: 10px; text-align: center;"&gt;
        &lt;tr&gt;
        &lt;th bgcolor="#AAAAAA"&gt;Tipo&lt;/th&gt;
        &lt;th bgcolor="#AAAAAA"&gt;Tensión Trafo Reemplazado&lt;/th&gt;
        &lt;th bgcolor="#AAAAAA"&gt;Tensión Trafo Nuevo&lt;/th&gt;
        &lt;th bgcolor="#AAAAAA"&gt;Capacidad&lt;/th&gt;
        &lt;/tr&gt;"""

    for i in range(n_trafos):
        estilo_base_trafo += f"""
        &lt;tr&gt;
        &lt;td&gt;$[{nombre}/trafo{i+1}_tipo]&lt;/td&gt;
        &lt;td&gt;$[{nombre}/trafo{i+1}_tension_trafo_reemplazado]&lt;/td&gt;
        &lt;td&gt;$[{nombre}/trafo{i+1}_tension_trafo_nvo]&lt;/td&gt;
        &lt;td&gt;$[{nombre}/trafo{i+1}_Capacidad]&lt;/td&gt;
        &lt;/tr&gt;"""

    estilo_base_trafo += """&lt;/table&gt;"""

    return estilo_base_trafo


def generar_parrafo_otros(nombre):

    largo_nombre = len(nombre)
    n_otros = int(nombre[largo_nombre-1])

    estilo_base_otros = """&lt;div style="margin-top: 10px; text-align: justify;"&gt;"""
    
    for i in range(n_otros):
        estilo_base_otros += f"""&lt;p&gt;$[{nombre}/parrafo{i+1}]&lt;/p&gt;"""

    estilo_base_otros += """&lt;/div&gt;"""

    return estilo_base_otros




def generar_estilo(nombre):
    # De la misma forma que para la funcion anterior, el parametro "nombre" tendra un codigo de numeros, donde el primer numero sera el numero de filas de la tabla de patios, el segundo el nro de filas de la tabla trafos y el tercero el numero de parrafos de la seccion otros.
    # A continuacion se muestra un ejemplo de como se generara el estilo para un proyecto con 1 patio, 1 trafo y 1 otros.

    largo_nombre = len(nombre)
    n_otros = int(nombre[largo_nombre-1])
    n_trafos = int(nombre[largo_nombre-2])
    n_patios = int(nombre[largo_nombre-3])


    print(f"n_patios: {n_patios}")
    print(f"n_trafos: {n_trafos}")
    print(f"n_otros: {n_otros}")

    #El id puede que tenga que anteponerle fol_nombre

    estilo_base = f"""
        <ns0:Style id="{nombre}">
            <ns0:IconStyle>
                <ns0:scale>1.16667</ns0:scale>
                <ns0:Icon>
                    <ns0:href>files/add_9.png</ns0:href>
                </ns0:Icon>
            </ns0:IconStyle>
            <ns0:LabelStyle>
                <ns0:color>00ffffff</ns0:color>
            </ns0:LabelStyle>
            <ns0:BalloonStyle>
                <ns0:text>&lt;table border=0 cellpadding=0 cellspacing=1 summary="" width=400&gt;
                    &lt;tr&gt;
                    &lt;td width="100%" valign="middle" align="center"&gt;
                    &lt;img src="files/trinergy_13.jpg" border="0" width="240" height="90"
                    alt=""&gt;
                    &lt;/td&gt;
                    &lt;/tr&gt;
                    &lt;tr cellpadding=1&gt;
                    &lt;td width="100%" valign="middle" align="center" bgcolor=#8bb145&gt;
                    &lt;b&gt;
                    &lt;font face="Arial" color="#ffffff"&gt;
                    &lt;h2.5&gt;NUEVA SUBESTACION&lt;/h2.5&gt;
                    &lt;/font&gt;
                    &lt;/b&gt;
                    &lt;/td&gt;
                    &lt;/tr&gt;
                    &lt;/table&gt;
                    &lt;table border="1" padding="0" style="margin-top: 10px; text-align: justify;
                    width:auto;"&gt;
                    &lt;tr&gt;&lt;td
                    bgcolor="#AAAAAA"&gt;&lt;b&gt;Nombre&lt;/b&gt;&lt;/td&gt;=&lt;td&gt;$[{nombre}/OBRA]&lt;/td&gt;&lt;/tr&gt;
                    &lt;tr&gt;&lt;td
                    bgcolor="#AAAAAA"&gt;&lt;b&gt;Decreto&lt;/b&gt;&lt;/td&gt;=&lt;td&gt;$[{nombre}/DECRETO]&lt;/td&gt;&lt;/tr&gt;
                    &lt;tr&gt;&lt;td
                    bgcolor="#AAAAAA"&gt;&lt;b&gt;Tipo&lt;/b&gt;&lt;/td&gt;=&lt;td&gt;$[{nombre}/TIPO]&lt;/td&gt;&lt;/tr&gt;
                    &lt;tr&gt;&lt;td bgcolor="#AAAAAA"&gt;&lt;b&gt;Valor
                    Inversion&lt;/b&gt;&lt;/td&gt;=&lt;td&gt;$[{nombre}/VI]&lt;/td&gt;&lt;/tr&gt;
                    &lt;tr&gt;&lt;td bgcolor="#AAAAAA"&gt;&lt;b&gt;Entrada en
                    Operacion&lt;/b&gt;&lt;/td&gt;=&lt;td&gt;$[{nombre}/ENTRADA_OP]&lt;/td&gt;&lt;/tr&gt;
                    &lt;tr&gt;
                    &lt;td bgcolor="#AAAAAA"&gt;&lt;b&gt;Resumen&lt;/b&gt;&lt;/td&gt;
                    &lt;td&gt;
                    &lt;div style="margin: 5px; padding-right: 10px; height:100px; overflow:auto;
                    text-align: justify;"&gt;
                    $[{nombre}/RESUMEN]
                    &lt;/div&gt;
                    &lt;/td&gt;
                    &lt;/tr&gt;
                    &lt;/table&gt;
                    """


    if n_patios != 0:
        estilo_base += generar_tabla_patios(nombre)

    if n_trafos != 0:
        estilo_base += generar_tabla_trafos(nombre)

    if n_otros != 0:
        estilo_base += generar_parrafo_otros(nombre)

    estilo_base += """</ns0:text>
    </ns0:BalloonStyle>
    </ns0:Style>"""
    
    return estilo_base


print(generar_estilo("Amp_200"))
```
